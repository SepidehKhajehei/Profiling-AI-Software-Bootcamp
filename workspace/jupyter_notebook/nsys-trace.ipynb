### NCCL AllReduce Trace

Before we start the trace, let's refresh our knowledge of the `Allreduce` operation. AllReduce allows all the nodes/GPUs to get the same aggregated value. The AllReduce function allows all workers to obtain the aggregated gradients from all worker nodes. This gradient aggregation is the model synchronization procedure in the AllReduce architecture. It guarantees that all the workers use the same gradient to update the model in the current training iteration. AllReduce uses all-to-all communication. Every worker needs to send their value to all the other workers. The AllReduce function is implemented as `Ring AllReduce` using NVIDIA NCCL. Since our DDP program was profiled on a single-node system with 4 GPUs, the screenshot below shows the NCCL kernel launch for the Ring AllReduce Sum operation using `FP32`. The tooltip on the kernel for `GPU0` detailed the sizes of shared memory, threads, and blocks, and also included latency and the total kernel execution time. It is important to note that AllReduce Kernels were launched by all the GPUs on the current timeline `(0.45s ~ 0.49s)` except for `GPU3`, which differs slightly. This occurrence may impact the `workers` synchronization process as they will wait for GPU3 in the `Ring AllReduce Sum operation` sync for all GPUs.

<br/>

GPU3 does run the AllReduce, just later: in this time window it is still finishing its backward pass. AllReduce is a collective operation, so no GPU can skip it. GPU0, GPU1, and GPU2 start their AllReduce kernels and then wait inside them until GPU3 joins, which is why their NCCL kernels look long. A GPU that holds up the others like this is called a *straggler*.

One way to trace the cause of the delay in GPU3 is to check if there was a delay in the data loader/data copy to the device (HtoD) and to look further into processes running on the CPU on the timeline. The data loading and pinned memory optimizations from the [previous notebook](nsys-application.ipynb) reduce exactly these delays.   

### NCCL Broadcast Trace

An AllReduce already leaves the aggregated gradients on every GPU, so no broadcast is needed for the gradients. The broadcast you see in every iteration comes from DDP keeping the model *buffers* in sync: by default (`broadcast_buffers=True`), rank 0 sends buffers such as the BatchNorm running mean and variance of ResNet to all other ranks at the start of each forward pass. The report screenshot below shows `GPU0` as the master worker performing `ncclBroadcast` task through the `ncclDevKernel_Broadcast_RING_LL` kernel. Meanwhile, GPU1 and GPU2 are busy on `memcpy HtoD` (copy from Host to device) while GPU3 performs a backward pass.

### NVLink Trace

Another important trace in a multi-GPU environment is the data transfer speed between devices. This process heavily relies on the NVLink bandwidth for transmission (Tx) and receiving (Rx), which is measured in throughput percentage. To check for NVLink Tx and Rx on the report, click on any GPUs. You will see the `GPU Metrics (10kHz or 5kHz)` click on it to drop-down, and there you will see the `NVLink Rx Bandwidth` and `NVLink Tx Bandwidth.` Hover your mouse on it to see the popup box that displays the `throughput %` as shown in the screenshot of the report for the NVLink transmission below.

### Stats System View

The Stats Systems View is a feature of Nsight Systems that provides summaries on time spent by CUDA Kernels, Memory Operations, GPU trace, etc. You can access this feature by switching from the default `Event View` to select `Stats Systems View` in the dropdown. From the top box in the report screenshot below, we deduce that `cudaStreamSynchronize` took 62% of the time of all the CUDA API calls. This implies that more operations that require synchronization frequently occur. Such a process could be allreduce-sum and broadcast Kernel calls. You may investigate further by checking the CUDA kernel summary to ascertain the assumption. In the bottom green box of the screenshot, We see that `ncclDevKernel_AllReduce_Sum_f32_RING_LL` executes frequently (73%), and this supports our assumption of frequent `cudaStreamSynchronize` call. More importantly, you can also verify the use of mixed-precision (FP16 & FP32) operations that were enabled in the previous notebook, as indicated by red arrows.
